# 02_dsp_feature_engineering.ipynb
DSP analysis and feature extraction.

```
import sys
sys.path.append('..')
import os
import numpy as np
from src.audio_processor import load_audio, get_fft, extract_mfcc
from src.utils import plot_fft
import librosa
```

## FFT Analysis

```
data_dir = '../data/raw'
files = [f for f in os.listdir(data_dir) if f.endswith('.wav')]
if files:
    y, sr = load_audio(os.path.join(data_dir, files[0]))
    freq, fft_values = get_fft(y, sr)
    plot_fft(freq, fft_values, title=f'FFT of {files[0]}')
```

## MFCC Extraction

```
if files:
    mfcc_features = extract_mfcc(y, sr)
    print(f'Extracted MFCC feature vector shape: {mfcc_features.shape}')
    print(f'Feature vector: {mfcc_features}')
```
