Exploration of audio signals.

```
import sys
sys.path.append('..')
import os
from src.audio_processor import load_audio
from src.utils import plot_waveform
import matplotlib.pyplot as plt
```

## Load Audio Files

```
data_dir = '../data/raw'
files = [f for f in os.listdir(data_dir) if f.endswith('.wav')]
print(f'Found {len(files)} audio files.')
```

## Visualize Waveforms

```
if files:
    y, sr = load_audio(os.path.join(data_dir, files[0]))
    plot_waveform(y, sr, title=f'Waveform of {files[0]}')
else:
    print('No wav files found in data/raw')
```
