# 04_inference.ipynb
Predict on new audio.

```
import sys
sys.path.append('..')
import os
from src.audio_processor import load_audio, extract_mfcc
from src.model_trainer import load_model
```

## Load Model

```
model_path = '../models/spu_model.pkl'
if os.path.exists(model_path):
    model = load_model(model_path)
    print('Model loaded successfully.')
else:
    print('Model file not found.')
```

## Predict on New Audio

```
data_dir = '../data/raw'
files = [f for f in os.listdir(data_dir) if f.endswith('.wav')]
if files and os.path.exists(model_path):
    y, sr = load_audio(os.path.join(data_dir, files[0]))
    features = extract_mfcc(y, sr).reshape(1, -1)
    prediction = model.predict(features)
    print(f'Predicted class for {files[0]}: {prediction[0]}')
```
