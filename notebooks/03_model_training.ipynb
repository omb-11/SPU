# 03_model_training.ipynb
Train the machine learning model.

```
import sys
sys.path.append('..')
import os
import numpy as np
from src.audio_processor import load_audio, extract_mfcc
from src.model_trainer import train_model, save_model
```

## Build Dataset
Note: For this example, we'll use the few files we have as dummy classes.

```
data_dir = '../data/raw'
files = [f for f in os.listdir(data_dir) if f.endswith('.wav')]
X = []
y = []

for i, file in enumerate(files):
    file_path = os.path.join(data_dir, file)
    y_audio, sr = load_audio(file_path)
    mfcc = extract_mfcc(y_audio, sr)
    X.append(mfcc)
    y.append(i % 2) # Dummy labels (0 or 1)

X = np.array(X)
y = np.array(y)
print(f'X shape: {X.shape}, y shape: {y.shape}')
```

## Train Model

```
if len(X) >= 2:
    model, accuracy, cm = train_model(X, y)
    print(f'Accuracy: {accuracy}')
    print(f'Confusion Matrix:\n{cm}')
    save_model(model, '../models/spu_model.pkl')
else:
    print('Not enough data to train (need at least 2 samples)')
```
